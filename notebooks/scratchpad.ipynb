```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from typing import Dict, Tuple, Annotated
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import make_column_transformer
import pandas as pd
import numpy as np
from faker import Faker
from datetime import datetime, timedelta
import random
import logging

```

```python
fake = Faker()
data = []

n=1000

logging.info("generating synthetic data...")

for _ in range(n):
        experience = random.randint(0, 20)
        education = random.choice(["Bachelors", "Masters", "PhD"])
        job = random.choice(["Data Scientist", "Software Engineer", "Analyst", "Account Executive"])
        industry = random.choice(["Tech", "Finance", "Healthcare", "Education"])
        location = random.choice(["San Francisco", "Austin", "New York", "London"])
        certs = random.randint(0, 5)
        created_at =  datetime.now()

        # base salary
        base = 30000 + experience * 2500 + certs * 1500

        # education bonus
        if education == "Masters":
            base += 10000
        elif education == "PhD":
            base += 20000

        # job multiplier
        job_multipliers = {
            "Data Scientist": 1.2,
            "Software Engineer": 1.15,
            "Account Executive": 1.10,
            "Analyst": 1.0
        }

        # location adjustment
        location_adjustment = {
            "San Francisco": 1.3,
            "New York": 1.2,
            "Austin": 1.1,
            "London": 1.15
        }

        salary = base * job_multipliers[job] * location_adjustment[location]

        data.append({
            "experience_years": experience,
            "education_level": education,
            "job_title": job,
            "industry": industry,
            "location": location,
            "certifications": certs,
            "salary": round(salary, 2),
            "created_at": created_at
        })

        logging.info(" synthetic data generated...")

        df = (pd.DataFrame(data))

```

```python
df['location'].unique()
```

```python
categorical_cols = ["location","job_title","industry"]
ordinal_cols = ["education_level"]

ohe=OneHotEncoder(handle_unknown='ignore',sparse_output=False).set_output(transform='pandas')
ode=OrdinalEncoder()

ct = make_column_transformer(
    (ohe, categorical_cols),
    (ode, ordinal_cols),   
    remainder="passthrough"
)
ct.set_output(transform="pandas")

```

```python
df_transformed = ct.fit_transform(df)
#df_transformed.rename(columns={"remainder__salary": "salary", 'remainder__created_at': 'created_at'}, inplace=True)
df_transformed.columns = [col.replace("remainder__", "") for col in df_transformed.columns]
print(df_transformed.columns)
```
