```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from typing import Dict, Tuple, Annotated
import pandas as pd
import numpy as np
from faker import Faker
from datetime import datetime, timedelta
import random
import logging

```

```python
def load_data() -> Tuple[
    # Notice we use a Tuple and Annotated to return 
    # multiple named outputs
    Annotated[pd.DataFrame, "X_train"],
    Annotated[pd.DataFrame, "X_test"],
    Annotated[pd.Series, "y_train"],
    Annotated[pd.Series, "y_test"],
]:
    logging.info("Loading data...")
    df = pd.read_csv("data/salary_dataset.csv")
    print(df.info())
    logging.info("Splitting into train and test...")
    y = df['salary']
    X = df[['experience']]
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, shuffle=True, random_state=42
    )
    return X_train, X_test, y_train, y_test
```

```python
X_train, X_test, y_train, y_test = load_data()

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

# Optional: peek at the data
X_train.head()
```

```python
def reg_trainer(
    X_train: pd.DataFrame,
    y_train: pd.Series,
) -> Tuple[
    Annotated[LinearRegression, "trained_model"],
    Annotated[float, "rmse"],
]:
    """Train a sklearn linear regression classifier."""

    model = LinearRegression()
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
 # Create a ZenML Model object
    #zenml_model = model(name="salary_prediction_model", model=model, metadata={"rmse": str(rmse)})

    return model, {"rmse": rmse}

```

```python
trained_model, rmse = reg_trainer(X_train=X_train, y_train=y_train)
print("RMSE:", rmse)
```
